## I. Phát biểu bài toán

Mục tiêu của dự án là phân tích dữ liệu bài viết tin tức nhằm khảo sát tính khả thi của việc xây dựng mô hình dự đoán **chuyên mục (category)** của bài báo dựa trên nội dung văn bản.

Cụ thể, mô hình cần dự đoán biến mục tiêu:

- **Y (Target variable):** chuyên mục của bài báo (category).
- **Xi (Features):** các đặc trưng được trích xuất từ **tiêu đề (title)** và **nội dung (content)** của bài viết.

Do biến mục tiêu **Y là biến danh mục**, bài toán được xác định là **bài toán phân lớp (classification)** trong học máy.



### Tính khả thi của bài toán

Bài toán được đánh giá là **khả thi** vì:

- Dữ liệu có **nhãn rõ ràng** (mỗi bài báo có một chuyên mục tương ứng).
- Nội dung văn bản chứa **các từ khóa đặc trưng cho từng chuyên mục** (ví dụ: sức khỏe, thể thao, kinh doanh,...).
- Tập dữ liệu có **kích thước đủ lớn**, giúp mô hình học được các mẫu ngôn ngữ đặc trưng.



### Các đặc trưng sử dụng trong mô hình

Các đặc trưng chính được sử dụng bao gồm:

- **Văn bản từ tiêu đề và nội dung bài viết**
- Văn bản được **tiền xử lý**:
  - chuyển chữ thường
  - loại bỏ ký tự đặc biệt và URL
  - tách từ tiếng Việt
  - loại bỏ stopwords
- Văn bản sau đó được **biểu diễn bằng TF-IDF vector** để đưa vào mô hình học máy.



### Kết luận

Bài toán dự đoán chuyên mục bài báo từ nội dung văn bản là **bài toán phân lớp văn bản (text classification)**.  
Dữ liệu có nhãn rõ ràng và chứa nhiều thông tin ngữ nghĩa, do đó **có thể sử dụng các phương pháp học máy để xây dựng mô hình dự đoán**.
