```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
DATA_PATH = "../data/"

df = pd.read_csv(DATA_PATH + "vnexpress_clean_data.csv")

print("Dataset shape:", df.shape)
df.head()
```

```python
df.info()
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
plt.figure()

sns.countplot(data=df, x="group")

plt.title("Articles per Group (Clean Data)")

plt.show()
```

```python
plt.figure()

sns.histplot(df["content_length"], bins=30)

plt.title("Content Length Distribution (Clean Data)")

plt.show()
```

```python
plt.figure()

sns.histplot(df["title_length"], bins=30)

plt.title("Title Length Distribution (Clean Data)")

plt.show()
```

```python
plt.figure()

sns.countplot(data=df, x="month")

plt.title("Articles by Month")

plt.show()
```

```python
plt.figure()

sns.countplot(data=df, x="hour")

plt.title("Articles by Hour")

plt.show()
```

```python
plt.figure()

sns.scatterplot(
    data=df,
    x="content_length",
    y="nums_of_comments",
    hue="group"
)

plt.title("Content Length vs Comments")

plt.show()
```
